```python
numbers = [12, 45, 23, 45, 67, 12]

largest = second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif largest > num > second_largest:
        second_largest = num

if second_largest == float('-inf'):
    print("No second largest unique element found.")
else:
    print("Second largest element:", second_largest)
```

```python
arr = [100, 4, 200, 1, 3, 2]

arr.sort()

count = 1
max_count = 1

for i in range(1, len(arr)):
    if arr[i] == arr[i - 1] + 1:
        count += 1
    elif arr[i] != arr[i - 1]:
        count = 1

    if count > max_count:
        max_count = count

print(max_count)
```

```python
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

print("Enter Matrix A:")
A = []
for i in range(r):
    row = list(map(int, input().split()))
    A.append(row)

print("Enter Matrix B:")
B = []
for i in range(r):
    row = list(map(int, input().split()))
    B.append(row)

print("\nAddition:")
for i in range(r):
    for j in range(c):
        print(A[i][j] + B[i][j], end=" ")
    print()

print("\nSubtraction:")
for i in range(r):
    for j in range(c):
        print(A[i][j] - B[i][j], end=" ")
    print()

print("\nMultiplication:")
for i in range(r):
    for j in range(c):
        s = 0
        for k in range(c):
            s += A[i][k] * B[k][j]
        print(s, end=" ")
    print()
```

```python
n = int(input("Enter size of matrix: "))

a = []
print("Enter matrix:")
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)

top = 0
bottom = n - 1
left = 0
right = n - 1

print("Spiral Order:")

while top <= bottom and left <= right:

    for i in range(left, right + 1):
        print(a[top][i], end=" ")
    top += 1

    for i in range(top, bottom + 1):
        print(a[i][right], end=" ")
    right -= 1

    if top <= bottom:
        for i in range(right, left - 1, -1):
            print(a[bottom][i], end=" ")
        bottom -= 1

    if left <= right:
        for i in range(bottom, top - 1, -1):
            print(a[i][left], end=" ")
        left += 1
```

```python
arr = list(map(int, input("Enter elements: ").split()))

n = len(arr)

for i in range(n):
    count = 0
    for j in range(n):
        if arr[i] == arr[j]:
            count += 1

    if count > n // 2:
        print("Majority Element:", arr[i])
        break
else:
    print("No Majority Element")
```

```python
prices = list(map(int, input("Enter stock prices: ").split()))

max_profit = 0

for i in range(len(prices)):
    for j in range(i + 1, len(prices)):
        profit = prices[j] - prices[i]
        if profit > max_profit:
            max_profit = profit

print("Maximum Profit:", max_profit)
```
